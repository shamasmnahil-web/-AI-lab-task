```python
import heapq

class AStar:
    def __init__(self):
        self.graph = {
            "S":[("A",1),("B",4)],
            "A":[("C",2)],
            "B":[("G",5)],
            "C":[("G",3)],
            "G":[]
        }
        self.h = {"S":5,"A":4,"B":4,"C":2,"G":0}
    def find_path(self, start, goal):
        q = [(self.h[start], 0, start, [start])]
        cost_so_far = {start:0}
        while q:
            f, cost, node, path = heapq.heappop(q)
            if cost != cost_so_far[node]:
                continue
            if node == goal:
                return path, cost
            for next_node, weight in self.graph[node]:
                new_cost = cost + weight

                if new_cost < cost_so_far.get(next_node, float("inf")):
                    cost_so_far[next_node] = new_cost
                    heapq.heappush(q,(new_cost + self.h[next_node], new_cost, next_node, path + [next_node]))
    def ucs(self, start, goal):
         q = [(0, start, start)]
         best = {start:0}

         while q:
             cost, node,path = heapq.heappop(q)

             if cost != best[node]:
                 continue
             if node == goal:
                 return path, cost
             for next_node, weight in self.graph[node]:
                 new_cost = cost + weight

                 if new_cost < best.get(next_node,float("inf")):
                     best[next_node] = new_cost
                     new_path = path + "->" + next_node
                     heapq.heappush(q,(new_cost, next_node, new_path))
a = AStar()
print("A*:",a.find_path("S","G"))
print("UCS:",a.ucs("S","G"))

a.h["C"] = 10
print("Overestimated:", a.find_path("S","G"))r

        
```
